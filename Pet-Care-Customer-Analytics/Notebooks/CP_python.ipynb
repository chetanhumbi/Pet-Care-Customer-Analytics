```python
import pandas as pd
from scipy.stats import chi2_contingency

# Categorical variable pairs we want to investigate
chi_square_pairs = [
    ("pet_has_active_subscription", "pet_food_tier"),
    ("pet_has_active_subscription", "pet_life_stage_at_order"),
    ("pet_has_active_subscription", "pet_breed_size"),
    ("pet_has_active_subscription", "gender"),
    ("pet_has_active_subscription", "neutered"),
    ("pet_food_tier", "pet_life_stage_at_order"),
    ("pet_food_tier", "pet_breed_size"),
    ("pet_life_stage_at_order", "pet_breed_size")
]

results = []

for col1, col2 in chi_square_pairs:
    
    # Remove missing values for the two variables
    temp = df[[col1, col2]].dropna()
    
    # Create contingency table
    table = pd.crosstab(temp[col1], temp[col2])
    
    # Chi-square test
    chi2, p_value, dof, expected = chi2_contingency(table)
    
    results.append({
        "Variable 1": col1,
        "Variable 2": col2,
        "Chi-Square": round(chi2, 2),
        "Degrees of Freedom": dof,
        "P-Value": p_value,
        "Significant": "Yes" if p_value < 0.05 else "No"
    })

chi_results = pd.DataFrame(results)

chi_results
```

```python
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    table = pd.crosstab(x, y)
    chi2, p, dof, expected = chi2_contingency(table)

    n = table.sum().sum()
    r, k = table.shape

    phi2 = chi2 / n

    # Bias correction
    phi2corr = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1))
    rcorr = r - ((r - 1) ** 2) / (n - 1)
    kcorr = k - ((k - 1) ** 2) / (n - 1)

    v = np.sqrt(phi2corr / min(kcorr - 1, rcorr - 1))

    return v


significant_pairs = [
    ("pet_has_active_subscription", "pet_food_tier"),
    ("pet_has_active_subscription", "pet_life_stage_at_order"),
    ("pet_has_active_subscription", "pet_breed_size"),
    ("pet_has_active_subscription", "neutered"),
    ("pet_food_tier", "pet_life_stage_at_order"),
    ("pet_food_tier", "pet_breed_size"),
    ("pet_life_stage_at_order", "pet_breed_size")
]

cramers_results = []

for col1, col2 in significant_pairs:
    temp = df[[col1, col2]].dropna()

    chi2, p, dof, expected = chi2_contingency(
        pd.crosstab(temp[col1], temp[col2])
    )

    v = cramers_v(temp[col1], temp[col2])

    cramers_results.append({
        "Variable 1": col1,
        "Variable 2": col2,
        "Chi-Square": round(chi2, 2),
        "P-Value": p,
        "Cramer's V": round(v, 3)
    })

cramers_results = pd.DataFrame(cramers_results)

cramers_results.sort_values("Cramer's V", ascending=False)
```

```python
from scipy.stats import levene

anova_tests = [
    ("total_minutes_on_website_since_last_order", "pet_food_tier"),
    ("total_minutes_on_website_since_last_order", "pet_life_stage_at_order"),
    ("total_minutes_on_website_since_last_order", "pet_has_active_subscription"),
    ("total_order_kcal", "pet_life_stage_at_order"),
    ("wet_trays", "pet_food_tier")
]

for numeric_col, group_col in anova_tests:
    
    print("\n" + "="*70)
    print(f"{numeric_col}  →  {group_col}")
    
    groups = [
        group[numeric_col].values
        for _, group in df[[numeric_col, group_col]]
        .dropna()
        .groupby(group_col)
    ]
    
    stat, p = levene(*groups)
    
    print(f"Levene statistic: {stat:.3f}")
    print(f"Levene p-value: {p:.6f}")
```

```python
from scipy.stats import f_oneway

welch_tests = [
    ("total_minutes_on_website_since_last_order", "pet_life_stage_at_order"),
    ("total_order_kcal", "pet_life_stage_at_order"),
    ("wet_trays", "pet_food_tier")
]

welch_results = []

for numeric_col, group_col in welch_tests:

    temp = df[[numeric_col, group_col]].dropna()

    groups = [
        group[numeric_col].values
        for _, group in temp.groupby(group_col)
    ]

    # Welch's ANOVA
    result = f_oneway(*groups, equal_var=False)

    welch_results.append({
        "Numerical Variable": numeric_col,
        "Grouping Variable": group_col,
        "Welch F-Statistic": round(result.statistic, 3),
        "P-Value": result.pvalue,
        "Significant": "Yes" if result.pvalue < 0.05 else "No"
    })

welch_results = pd.DataFrame(welch_results)

welch_results
```

```python
def eta_squared(df, numeric_col, group_col):
    
    temp = df[[numeric_col, group_col]].dropna()
    
    grand_mean = temp[numeric_col].mean()
    
    # Between-group sum of squares
    group_stats = temp.groupby(group_col)[numeric_col].agg(["mean", "count"])
    
    ss_between = (
        group_stats["count"] *
        (group_stats["mean"] - grand_mean) ** 2
    ).sum()
    
    # Total sum of squares
    ss_total = (
        (temp[numeric_col] - grand_mean) ** 2
    ).sum()
    
    return ss_between / ss_total


effect_tests = [
    ("total_minutes_on_website_since_last_order", "pet_life_stage_at_order"),
    ("total_order_kcal", "pet_life_stage_at_order"),
    ("wet_trays", "pet_food_tier")
]

effect_results = []

for numeric_col, group_col in effect_tests:
    
    eta2 = eta_squared(df, numeric_col, group_col)
    
    effect_results.append({
        "Numerical Variable": numeric_col,
        "Grouping Variable": group_col,
        "Eta Squared": round(eta2, 4)
    })

effect_results = pd.DataFrame(effect_results)

effect_results
```

```python
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind
from itertools import combinations
from statsmodels.stats.multitest import multipletests

numeric_col = "total_order_kcal"
group_col = "pet_life_stage_at_order"

temp = df[[numeric_col, group_col]].dropna()

groups = temp[group_col].unique()

posthoc_results = []

for group1, group2 in combinations(groups, 2):

    x = temp[temp[group_col] == group1][numeric_col]
    y = temp[temp[group_col] == group2][numeric_col]

    # Welch's independent t-test
    t_stat, p_value = ttest_ind(
        x,
        y,
        equal_var=False
    )

    mean_diff = x.mean() - y.mean()

    posthoc_results.append({
        "Group 1": group1,
        "Group 2": group2,
        "Mean Group 1": x.mean(),
        "Mean Group 2": y.mean(),
        "Mean Difference": mean_diff,
        "Welch t-statistic": t_stat,
        "Raw p-value": p_value
    })

posthoc_results = pd.DataFrame(posthoc_results)

# Holm correction for multiple comparisons
reject, corrected_p, _, _ = multipletests(
    posthoc_results["Raw p-value"],
    method="holm"
)

posthoc_results["Adjusted p-value"] = corrected_p
posthoc_results["Significant"] = np.where(
    reject, "Yes", "No"
)

posthoc_results = posthoc_results.round({
    "Mean Group 1": 2,
    "Mean Group 2": 2,
    "Mean Difference": 2,
    "Welch t-statistic": 3,
    "Raw p-value": 6,
    "Adjusted p-value": 6
})

posthoc_results
```
